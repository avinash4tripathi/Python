List: List is collection of Home, Heterogenous data store inside the square braces []



List is mutable in nature where we can add,update,remove data from the collection.

```python
a = [1,2,3,4,5]
a.append('hello') # add the value at the end
print(a)
```

```python
a = [1,2,3,4.5]
a.insert(3,25) # Add the value at the index_postion 
print(a)
```

```python
a = [1,2,3,4,5,6]
a.pop(5)
print(a)
```

Tuple: It is a collection of homo, heterogamous data stored inside the parenthesis In python ()

It is known as most secure data type used for security. Where we can access the data but can’t modify it.
Tuple are immutable in nature.
Tuple is most secure datatype in python.


```python
a = ("apple","Mango","Graps")
print(a)
print(a[0])
print(a[1])
```

```python
a = ("apple","Mango","Graps")
#a[0] = "Orange"
print(a)
```
