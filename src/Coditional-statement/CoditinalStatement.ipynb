```python
# WAP to check the Character is in Uppercase?
Char = input("enter the character: ")
if 'A'<=Char<='Z':
    print("Uppercase")
else:
    print("lowecase")
```

```python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid day")
```
